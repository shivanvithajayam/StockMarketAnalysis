```python
import pandas as pd
import os
import glob

# Folder paths
raw_folder = "./data/raw"
clean_folder = "./data/cleaned"

os.makedirs(clean_folder, exist_ok=True)

# Read all 10 company delivery files
files = glob.glob(raw_folder + "/*-EQ-N.csv")

all_stocks = []

for file in files:

    df = pd.read_csv(file)

    # Remove extra spaces from column names
    df.columns = df.columns.str.strip()

    # Convert date
    df["Date"] = pd.to_datetime(
        df["Date"],
        format="%d-%b-%Y",
        errors="coerce"
    )

    # Clean numeric columns
    numeric_cols = [
        "Prev Close",
        "Open Price",
        "High Price",
        "Low Price",
        "Last Price",
        "Close Price",
        "Average Price",
        "Total Traded Quantity",
        "Turnover ₹",
        "No. of Trades",
        "Deliverable Qty",
        "% Dly Qt to Traded Qty"
    ]

    for col in numeric_cols:
        df[col] = pd.to_numeric(
            df[col].astype(str)
            .str.replace(",", "", regex=False)
            .str.replace("₹", "", regex=False)
            .str.strip(),
            errors="coerce"
        )

    # Remove invalid rows
    df = df.dropna(subset=["Date", "Close Price"])

    # Sort dates
    df = df.sort_values("Date")

    # Calculate daily percentage return
    df["Daily Return (%)"] = (
        df["Close Price"].pct_change() * 100
    )

    all_stocks.append(df)

# Combine all companies
combined_df = pd.concat(all_stocks, ignore_index=True)

# Save cleaned data
combined_df.to_csv(
    clean_folder + "/all_stocks_cleaned.csv",
    index=False
)

print("DATA CLEANING COMPLETED!")
print("Total companies:", combined_df["Symbol"].nunique())
print("Total records:", len(combined_df))
print("Companies:", combined_df["Symbol"].unique())

print("\nMissing values:")
print(combined_df.isnull().sum())

print("\nFirst 5 rows:")
display(combined_df.head())
```

```python
# Load the cleaned NIFTY 50 dataset

nifty = pd.read_csv(clean_folder + "/nifty50_cleaned.csv")

# Convert Date columns into datetime format
stocks["Date"] = pd.to_datetime(stocks["Date"])
nifty["Date"] = pd.to_datetime(nifty["Date"])

# Calculate and compare the performance of each stock with NIFTY 50

for company in stocks["Symbol"].unique():

    # Select data for the current company
    company_data = stocks[stocks["Symbol"] == company][
        ["Date", "Close Price"]
    ].copy()

    # Select NIFTY 50 date and closing price
    nifty_data = nifty[["Date", "Close"]].copy()

    # Merge both datasets using common dates
    comparison = pd.merge(
        company_data,
        nifty_data,
        on="Date",
        how="inner"
    )

    # Sort data by date
    comparison = comparison.sort_values("Date")

    # Convert both prices into Base 100 values
    comparison["Stock Performance"] = (
        comparison["Close Price"] / comparison["Close Price"].iloc[0]
    ) * 100

    comparison["NIFTY Performance"] = (
        comparison["Close"] / comparison["Close"].iloc[0]
    ) * 100

    # Plot stock performance against NIFTY 50
    plt.figure(figsize=(12, 5))

    plt.plot(
        comparison["Date"],
        comparison["Stock Performance"],
        label=company
    )

    plt.plot(
        comparison["Date"],
        comparison["NIFTY Performance"],
        label="NIFTY 50"
    )

    plt.title(company + " vs NIFTY 50 - Performance Comparison")
    plt.xlabel("Date")
    plt.ylabel("Indexed Performance (Base 100)")

    plt.legend()
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Arrange stock daily returns into separate columns
stock_returns = stocks.pivot_table(
    index="Date",
    columns="Symbol",
    values="Daily Return (%)"
)

# Select NIFTY 50 daily returns
nifty_returns = nifty[["Date", "Daily Return (%)"]].copy()

# Rename the NIFTY return column
nifty_returns = nifty_returns.rename(
    columns={"Daily Return (%)": "NIFTY 50"}
)

# Set Date as index
nifty_returns = nifty_returns.set_index("Date")

# Combine stock returns with NIFTY 50 returns
returns = stock_returns.join(nifty_returns, how="inner")

# Calculate correlation between all companies
correlation_matrix = returns.corr()

# Display correlation heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Stock and NIFTY 50 Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
# Calculate total return for each company

total_returns = {}

for company in stocks["Symbol"].unique():

    # Get data for the current company
    company_data = stocks[stocks["Symbol"] == company].sort_values("Date")

    # Get first and last closing prices
    first_price = company_data["Close Price"].iloc[0]
    last_price = company_data["Close Price"].iloc[-1]

    # Calculate total return percentage
    total_return = ((last_price - first_price) / first_price) * 100

    total_returns[company] = total_return

# Convert results into a DataFrame
returns_df = pd.DataFrame(
    list(total_returns.items()),
    columns=["Company", "Total Return (%)"]
)

# Sort companies by total return
returns_df = returns_df.sort_values(
    by="Total Return (%)",
    ascending=False
)

# Display the results
print("STOCK PERFORMANCE RANKING")
print(returns_df)

# Plot the total returns
plt.figure(figsize=(12, 5))

plt.bar(
    returns_df["Company"],
    returns_df["Total Return (%)"],
    edgecolor="black"
)

plt.title("Top and Bottom Performing Stocks")
plt.xlabel("Company")
plt.ylabel("Total Return (%)")
plt.xticks(rotation=45)
plt.axhline(0, color="black")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

```python
# Calculate average annualized return and annualized risk

risk_return = stocks.groupby("Symbol").agg(
    Average_Daily_Return=("Daily Return (%)", "mean"),
    Daily_Volatility=("Daily Return (%)", "std")
)

# Convert daily return into annualized return
risk_return["Annualized Return (%)"] = (
    risk_return["Average_Daily_Return"] * 252
)

# Convert daily volatility into annualized risk
risk_return["Annualized Risk (%)"] = (
    risk_return["Daily_Volatility"] * (252 ** 0.5)
)

# Create a scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(
    risk_return["Annualized Risk (%)"],
    risk_return["Annualized Return (%)"],
    s=100
)

# Add company names beside each point
for company in risk_return.index:
    plt.annotate(
        company,
        (
            risk_return.loc[company, "Annualized Risk (%)"],
            risk_return.loc[company, "Annualized Return (%)"]
        )
    )

# Add graph title and axis labels
plt.title("Risk vs Return Analysis")
plt.xlabel("Annualized Risk / Volatility (%)")
plt.ylabel("Annualized Return (%)")

plt.axhline(0, color="black")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Calculate 20-day and 50-day moving averages for each company

stocks = stocks.sort_values(["Symbol", "Date"]).copy()

stocks["20 Day MA"] = stocks.groupby("Symbol")[
    "Close Price"
].transform(lambda x: x.rolling(window=20).mean())

stocks["50 Day MA"] = stocks.groupby("Symbol")[
    "Close Price"
].transform(lambda x: x.rolling(window=50).mean())

# Plot moving averages for each company

for company in stocks["Symbol"].unique():

    company_data = stocks[stocks["Symbol"] == company]

    plt.figure(figsize=(12, 5))

    # Plot actual closing price
    plt.plot(
        company_data["Date"],
        company_data["Close Price"],
        label="Closing Price",
        alpha=0.5
    )

    # Plot 20-day moving average
    plt.plot(
        company_data["Date"],
        company_data["20 Day MA"],
        label="20-Day Moving Average"
    )

    # Plot 50-day moving average
    plt.plot(
        company_data["Date"],
        company_data["50 Day MA"],
        label="50-Day Moving Average"
    )

    plt.title(company + " - Moving Average Analysis")
    plt.xlabel("Date")
    plt.ylabel("Closing Price (₹)")

    plt.legend()
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Calculate daily returns of each stock and NIFTY 50

stock_returns = stocks.pivot_table(
    index="Date",
    columns="Symbol",
    values="Daily Return (%)"
)

nifty_returns = nifty.set_index("Date")["Daily Return (%)"]

# Combine stock returns and NIFTY 50 returns
returns = stock_returns.join(
    nifty_returns.rename("NIFTY 50"),
    how="inner"
)

# Calculate correlation of each stock with NIFTY 50
nifty_correlation = returns.corr()["NIFTY 50"].drop("NIFTY 50")

# Sort correlations
nifty_correlation = nifty_correlation.sort_values(ascending=False)

# Display correlation values
print("STOCK VS NIFTY 50 CORRELATION")
print(nifty_correlation)

# Plot correlation values
plt.figure(figsize=(12, 5))

plt.bar(
    nifty_correlation.index,
    nifty_correlation.values,
    edgecolor="black"
)

plt.title("Correlation of Individual Stocks with NIFTY 50")
plt.xlabel("Company")
plt.ylabel("Correlation")

plt.axhline(0, color="black")
plt.grid(axis="y")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Calculate 20-day rolling volatility for each company

stocks = stocks.sort_values(["Symbol", "Date"]).copy()

stocks["Rolling Volatility"] = stocks.groupby("Symbol")[
    "Daily Return (%)"
].transform(
    lambda x: x.rolling(window=20).std() * (252 ** 0.5)
)

# Plot rolling volatility for each company

for company in stocks["Symbol"].unique():

    # Filter data for the current company
    company_data = stocks[stocks["Symbol"] == company]

    # Create a new figure
    plt.figure(figsize=(12, 5))

    # Plot rolling volatility
    plt.plot(
        company_data["Date"],
        company_data["Rolling Volatility"],
        label=company
    )

    # Add graph title and axis labels
    plt.title(company + " - 20 Day Rolling Volatility")
    plt.xlabel("Date")
    plt.ylabel("Annualized Volatility (%)")

    # Display legend and grid
    plt.legend()
    plt.grid(True)

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Calculate maximum drawdown for each company

max_drawdowns = {}

for company in stocks["Symbol"].unique():

    # Get data for the current company
    company_data = stocks[
        stocks["Symbol"] == company
    ].sort_values("Date").copy()

    # Calculate cumulative stock performance
    cumulative_returns = (
        1 + company_data["Daily Return (%)"].fillna(0) / 100
    ).cumprod()

    # Calculate previous highest value
    running_peak = cumulative_returns.cummax()

    # Calculate drawdown percentage
    drawdown = (
        (cumulative_returns - running_peak) / running_peak
    ) * 100

    # Store the maximum drawdown
    max_drawdowns[company] = drawdown.min()

# Convert results into a DataFrame
drawdown_df = pd.DataFrame(
    list(max_drawdowns.items()),
    columns=["Company", "Maximum Drawdown (%)"]
)

# Sort companies by maximum drawdown
drawdown_df = drawdown_df.sort_values(
    by="Maximum Drawdown (%)"
)

# Display results
print("MAXIMUM DRAWDOWN COMPARISON")
print(drawdown_df)

# Plot maximum drawdown
plt.figure(figsize=(12, 5))

plt.bar(
    drawdown_df["Company"],
    drawdown_df["Maximum Drawdown (%)"],
    edgecolor="black"
)

plt.title("Maximum Drawdown Comparison")
plt.xlabel("Company")
plt.ylabel("Maximum Drawdown (%)")

plt.xticks(rotation=45)
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

```python
# Create a final comparison table for all companies

final_analysis = stocks.groupby("Symbol").agg(
    Average_Daily_Return=("Daily Return (%)", "mean"),
    Daily_Volatility=("Daily Return (%)", "std"),
    Average_Delivery=("% Dly Qt to Traded Qty", "mean"),
    Average_Volume=("Total Traded Quantity", "mean")
)

# Calculate annualized return and volatility
final_analysis["Annualized Return (%)"] = (
    final_analysis["Average_Daily_Return"] * 252
)

final_analysis["Annualized Volatility (%)"] = (
    final_analysis["Daily_Volatility"] * (252 ** 0.5)
)

# Calculate total return
total_return = stocks.groupby("Symbol").apply(
    lambda x: (
        (x.sort_values("Date")["Close Price"].iloc[-1] /
         x.sort_values("Date")["Close Price"].iloc[0]) - 1
    ) * 100
)

final_analysis["Total Return (%)"] = total_return

# Add maximum drawdown
final_analysis["Maximum Drawdown (%)"] = pd.Series(max_drawdowns)

# Select important comparison columns
final_analysis = final_analysis[
    [
        "Total Return (%)",
        "Annualized Volatility (%)",
        "Average_Delivery",
        "Average_Volume",
        "Maximum Drawdown (%)"
    ]
]

# Display final comparison table
print("FINAL STOCK COMPARISON")
print(final_analysis.round(2))
```

```python
# Load the cleaned NIFTY 50 dataset

nifty = pd.read_csv(clean_folder + "/nifty50_cleaned.csv")

# Convert Date column into datetime format
nifty["Date"] = pd.to_datetime(nifty["Date"])

# Sort data by date
nifty = nifty.sort_values("Date").copy()

# Calculate 20-day rolling volatility
nifty["Rolling Volatility"] = (
    nifty["Daily Return (%)"]
    .rolling(window=20)
    .std() * (252 ** 0.5)
)

# Calculate absolute daily returns
nifty["Absolute Daily Return"] = (
    nifty["Daily Return (%)"].abs()
)

# Calculate correlation between market movement and volatility
correlation = nifty["Absolute Daily Return"].corr(
    nifty["Rolling Volatility"]
)

# Display correlation
print("NIFTY 50 VOLATILITY VS INDEX MOVEMENT")
print("Correlation:", round(correlation, 3))

# Plot NIFTY 50 daily returns and rolling volatility
fig, ax1 = plt.subplots(figsize=(12, 5))

# Plot absolute daily returns
ax1.plot(
    nifty["Date"],
    nifty["Absolute Daily Return"],
    label="Absolute Daily Return"
)

ax1.set_xlabel("Date")
ax1.set_ylabel("Absolute Daily Return (%)")

# Create a second y-axis
ax2 = ax1.twinx()

# Plot rolling volatility
ax2.plot(
    nifty["Date"],
    nifty["Rolling Volatility"],
    label="20-Day Rolling Volatility"
)

ax2.set_ylabel("Annualized Volatility (%)")

# Add graph title
plt.title("NIFTY 50 - Volatility vs Index Movement")

# Display legends
ax1.legend(loc="upper left")
ax2.legend(loc="upper right")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
